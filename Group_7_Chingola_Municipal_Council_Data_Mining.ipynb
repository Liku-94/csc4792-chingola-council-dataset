````python
all_files = []

for root, directories, files_in_folder in os.walk(extract_dir):
    for file in files_in_folder:
        all_files.append(os.path.join(root, file))

print(f"Total files found: {len(all_files)}\n")

for file in all_files:
    print("-", file)
````

````python
csv_files = [
    file for file in all_files
    if file.lower().endswith(".csv")
]

print(f"Total CSV files: {len(csv_files)}")

for file in csv_files:
    print("-", os.path.basename(file))
````

````python
raw_data = {}

for file in csv_files:

    filename = os.path.basename(file)

    try:
        df = pd.read_csv(
            file,
            sep="|",
            encoding="utf-8"
        )

        raw_data[filename] = df

        print(
            f"{filename}: "
            f"{df.shape[0]} rows × {df.shape[1]} columns"
        )

    except Exception as e:
        print(f"ERROR reading {filename}: {e}")
````

````python
dataset_inventory = []

for filename, df in raw_data.items():

    dataset_inventory.append({
        "Dataset": filename,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "MissingValues": int(df.isna().sum().sum()),
        "DuplicateRows": int(df.duplicated().sum())
    })

inventory_df = pd.DataFrame(dataset_inventory)

inventory_df = inventory_df.sort_values(
    by="Rows",
    ascending=False
)

display(inventory_df)
````

````python
for filename, df in raw_data.items():

    print("\n" + "=" * 100)
    print(filename)
    print("=" * 100)

    print("Shape:", df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nFirst 3 records:")
    display(df.head(3))
````

````python
quality_reports = []

for filename, df in raw_data.items():

    for column in df.columns:

        quality_reports.append({
            "Dataset": filename,
            "Column": column,
            "DataType": str(df[column].dtype),
            "MissingValues": int(df[column].isna().sum()),
            "MissingPercent": round(
                df[column].isna().mean() * 100, 2
            ),
            "UniqueValues": int(df[column].nunique()),
            "DuplicateValues": int(
                df[column].duplicated().sum()
            )
        })

quality_df = pd.DataFrame(quality_reports)

display(quality_df)
````

````python
duplicate_report = []

for filename, df in raw_data.items():

    duplicate_report.append({
        "Dataset": filename,
        "Rows": len(df),
        "DuplicateRows": int(df.duplicated().sum())
    })

duplicate_df = pd.DataFrame(duplicate_report)

display(duplicate_df)
````

````python
cleaned_data = {}

def standardise_column_name(column):

    column = str(column).strip()

    column = re.sub(
        r"(?<!^)(?=[A-Z])",
        "_",
        column
    )

    column = column.lower()

    column = re.sub(
        r"[^a-z0-9]+",
        "_",
        column
    )

    column = column.strip("_")

    return column


for filename, df in raw_data.items():

    clean_df = df.copy()

    clean_df.columns = [
        standardise_column_name(col)
        for col in clean_df.columns
    ]

    cleaned_data[filename] = clean_df

print("Column names standardised successfully.")
````

````python
for filename, df in cleaned_data.items():

    text_columns = df.select_dtypes(
        include=["object"]
    ).columns

    for column in text_columns:

        df[column] = (
            df[column]
            .astype("string")
            .str.strip()
        )

        # Convert empty strings to missing values
        df[column] = df[column].replace(
            {"": pd.NA, "nan": pd.NA, "None": pd.NA}
        )

print("Text cleaning completed.")
````

````python
duplicate_removal = []

for filename, df in cleaned_data.items():

    before = len(df)

    df.drop_duplicates(
        inplace=True
    )

    after = len(df)

    duplicate_removal.append({
        "Dataset": filename,
        "RowsBefore": before,
        "RowsAfter": after,
        "DuplicatesRemoved": before - after
    })

duplicate_removal_df = pd.DataFrame(
    duplicate_removal
)

display(duplicate_removal_df)
````

````python
def convert_numeric_columns(df):

    df = df.copy()

    for column in df.columns:

        if df[column].dtype == "object":

            sample = (
                df[column]
                .dropna()
                .astype(str)
                .str.replace(",", "", regex=False)
                .str.replace("K", "", regex=False)
                .str.replace("ZMW", "", regex=False)
                .str.replace("ZMK", "", regex=False)
                .str.strip()
            )

            numeric_version = pd.to_numeric(
                sample,
                errors="coerce"
            )

            if (
                len(sample) > 0
                and numeric_version.notna().mean() >= 0.80
            ):
                df[column] = pd.to_numeric(
                    df[column]
                    .astype(str)
                    .str.replace(",", "", regex=False)
                    .str.replace("K", "", regex=False)
                    .str.replace("ZMW", "", regex=False)
                    .str.replace("ZMK", "", regex=False)
                    .str.strip(),
                    errors="coerce"
                )

    return df


for filename in cleaned_data:

    cleaned_data[filename] = convert_numeric_columns(
        cleaned_data[filename]
    )

print("Numeric conversion completed.")
````

````python
missing_before = []

for filename, df in cleaned_data.items():

    missing_before.append({
        "Dataset": filename,
        "MissingValues": int(df.isna().sum().sum())
    })

missing_before_df = pd.DataFrame(missing_before)

display(missing_before_df)
````
